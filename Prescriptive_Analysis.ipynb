```python
#define costs
supplier_cost = 3
retail_price = 5

#starting values to find the stock with the highest profit
optimal_profit = -float('inf')
optimal_stock = 0

#analyse possible stock levels
for stock in range(30, 51):
    total_profit = 0

    #simulate different demand levels
    for demand in range(30,51):
        units_sold = min(stock, demand)
        profit = (units_sold * retail_price) - (stock * supplier_cost)
        total_profit += profit

    average_profit = total_profit / 21

    if average_profit > optimal_profit:
        optimal_profit = average_profit
        optimal_stock = stock

print(f"Optimal stock level: {optimal_stock} units")
print(f"Estimated average profit: £{optimal_profit:.2f}")
```
